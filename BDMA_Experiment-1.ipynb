```python
le=LabelEncoder()
```

```python
a['sex']=le.fit_transform(a['sex'])
```

```python
a['embark_town']=le.fit_transform(a['embark_town'])
```

```python
scaler=StandardScaler()
```

```python
n_c=['age','fare']
```

```python
a[n_c]=scaler.fit_transform(a[n_c])
```
